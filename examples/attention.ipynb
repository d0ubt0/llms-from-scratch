```python
import torch
```

```python
torch.manual_seed(42)

dim = 3

inputs_emb = torch.tensor(
[[0.43, 0.15, 0.89], # Your

[0.55, 0.87, 0.66], # journey

[0.57, 0.85, 0.64], # starts

[0.22, 0.58, 0.33], # with

[0.77, 0.25, 0.10], # one

[0.05, 0.80, 0.55]] # step

)
```

```python
query = inputs_emb[1]

attention_x2 = torch.empty(inputs_emb.shape[0])

for i in range(len(inputs_emb)):
    attention_x2[i] = inputs_emb[i].dot(query)

soft = torch.nn.Softmax(dim=0)

attention_x2_norm = soft(attention_x2)
attention_x2_norm
```

```python
context_vec = torch.zeros(query.shape)

for i in range(len(inputs_emb)):
    context_vec += attention_x2_norm[i] * inputs_emb[i]

context_vec
```

```python
attention_scores = inputs_emb @ inputs_emb.T
attention_scores
```

```python
inputs_emb
```

```python
attention_weight = torch.softmax(attention_scores, dim = 1)
attention_weight
```

```python
all_context = attention_weight @ inputs_emb
all_context
```
